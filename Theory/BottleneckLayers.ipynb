### A note on Bottleneck Layers

**Bottleneck Convolution Layers:** in the ResNet-50 architecture, the block consists of three convolutional layers w/ batch normalization and ReLU activation after each

- The first convolutional layer uses a filter size of `1x1` and reduces the number of channels in the input data. This dimensionality reduction helps to compress the data and improve computational efficiency without sacrificing too much information.

- The second convolutional layer uses a filter size of `3x3` to extract complex spatial features from the data.

- The third convolutional layer again uses a filter size of `1x1` to restore the original number of channels before the output is added to the shortcut connection.

- - -
### Why are they called Bottleneck layers?

**To give you an intuition:** they are called that because they act like the narrow neck of a bottle, restricting the flow of information by forcing it through a dramatically smaller channel capacity

It is designed to temporarily squeeze high-dimensional data into a much lower-dimensional space before expanding it back out

- - -

- - -
### The Core Benefits of Squeezing the Data

- **Computational Efficiency:** A standard 3x3 convolution running on a massive number of channels is incredibly expensive. By using a 1x1 convolution to temporarily bottleneck the channels down, the expensive 3x3 convolution is performed on a much smaller volume of data. This dramatically reduces training time, memory consumption, and floating-point operations (FLOPs).

- **Feature Compression:** Forcing data through a tight space acts as a natural filter. It forces the network to strip away background noise and redundant details, keeping only the most vital, high-level information needed to pass through the tight gap
